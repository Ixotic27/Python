# <span style="color:#2E86C1;">✨ Explain Function Parameters in Python (Positional, Keyword, Default, Variable-Length: *args, **kwargs)</span>
This notebook explains the four types of function parameters in Python with clear theory and ONE merged code block.

---

# <span style="color:#AF7AC5;">📌 What Are Function Parameters?</span>
Function parameters decide **how values are passed** into a function.  
Python supports four main types: **Positional**, **Keyword**, **Default**, and **Variable-Length**.

---

# <span style="color:#AF7AC5;">📌 All Parameters Explanation + Merged Example Code</span>

```python
# -----------------------------------------
# 1️⃣ Positional Parameters
# -----------------------------------------
# Values are assigned based on order.

def add(a, b):
    return a + b

print("Positional:", add(5, 10))   # a=5, b=10


# -----------------------------------------
# 2️⃣ Keyword Parameters
# -----------------------------------------
# Arguments passed using their names.

def info(name, age):
    return f"{name} is {age} years old"

print("Keyword:", info(age=20, name="Ishant"))


# -----------------------------------------
# 3️⃣ Default Parameters
# -----------------------------------------
# Parameter has a preset value. Used if no value is passed.

def greet(name, city="Dehradun"):
    return f"Hello {name} from {city}"

print("Default 1:", greet("Ishant"))          # uses default city
print("Default 2:", greet("Ishant", "Delhi")) # overrides default


# -----------------------------------------
# 4️⃣ Variable-Length Parameters
# -----------------------------------------
# *args → collects extra positional arguments as a TUPLE
# **kwargs → collects extra keyword arguments as a DICTIONARY

def show_args(*args):
    print("ARGS:", args)          # tuple of arguments

def show_kwargs(**kwargs):
    print("KWARGS:", kwargs)      # dictionary of arguments

show_args(1, 2, 3, 4)
show_kwargs(name="Ishant", age=20, city="Dehradun")

# <span style="color:#AF7AC5;">Tuple (`tuple`)</span>
Tuples are **ordered, immutable** collections of elements.  
They are created using parentheses `( )`.  
Since tuples **cannot be changed** after creation, they are used to store **fixed or protected data**.  
They allow duplicates and support indexing, slicing, and mixed data types.

---

### 🔹 When to Use Tuples?
✔ When data should **not change** (days, months, coordinates)  
✔ When **faster performance** is required  
✔ When using tuples as **dictionary keys** (immutable)  
✔ When **returning multiple values** from a function  
✔ When you want **memory-efficient** and safe data storage  

---

### 🔹 Example & 5 Functions:
```python
t = (10, 20, 30, 20, 50)

print(t.count(20))   # count occurrences
print(t.index(30))   # find index of value
print(len(t))        # tuple size
print(max(t))        # maximum value
print(min(t))        # minimum value

---

# <span style="color:#AF7AC5;">Dictionary (`dict`)</span>
A dictionary in Python stores data in **key–value pairs**.  
It is **unordered**, **mutable**, and keys must be **unique** and **immutable**.  
Dictionaries allow fast access and modification, making them ideal for structured data.

---

### 🔹 Uses of Dictionaries
✔ When storing data as **key → value** (e.g., name → phone)  
✔ When fast **search, insert, update** operations are needed  
✔ For **JSON-like** structured data  
✔ Used in **APIs, databases, ML models, configs**  
✔ Ideal for storing object-like information (student, employee, etc.)

---

### 🔹 Example & Operations (Create, Update, Delete)
```python
# -------------------------
# CREATE a Dictionary
# -------------------------
student = {
    "name": "Ishant",
    "age": 20,
    "course": "CSE"
}

print(student)   # {'name': 'Ishant', 'age': 20, 'course': 'CSE'}


# -------------------------
# UPDATE Dictionary
# -------------------------
student["age"] = 21               # update existing value
student["city"] = "Dehradun"      # add new key-value pair
student.update({"course": "AI"})  # update using update()

print(student)
# {'name': 'Ishant', 'age': 21, 'course': 'AI', 'city': 'Dehradun'}


# -------------------------
# DELETE from Dictionary
# -------------------------
student.pop("city")     # delete a key-value pair
del student["age"]      # delete using 'del'
student.clear()         # remove all items

print(student)          # {}

# <span style="color:#AF7AC5;">Set (`set`)</span>
A **set** in Python is an unordered collection of **unique** elements.  
Sets automatically remove duplicates and do not support indexing or slicing.  
They are mainly used for mathematical set operations.

---

### 🔹 Uses of Sets
✔ When duplicate values should be **automatically removed**  
✔ For performing **union, intersection, difference** operations  
✔ When checking **membership** quickly (`in` is very fast)  
✔ When storing **unordered, unique data**  
✔ Ideal for tasks like removing duplicates from a list  

---

### 🔹 Example & 5 Functions:
```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

s1.add(10)                 # add an element
s1.remove(2)               # remove an element
print(s1.union(s2))        # union of sets
print(s1.intersection(s2)) # common elements
print(s1.difference(s2))   # elements in s1 but not in s2


````python
d = {"a": 1, "b": 2, "c": 3}

print(d.keys())        # 1️⃣ keys(): returns all keys
print(d.values())      # 2️⃣ values(): returns all values
print(d.items())       # 3️⃣ items(): returns key-value pairs
print(d.get("b"))      # 4️⃣ get(): fetch value safely
print(d.popitem())     # 5️⃣ popitem(): removes last inserted pair
````

# <span style="color:#AF7AC5;">📌 What is File Handling?</span>
File handling in Python allows a program to **store data permanently** on the disk.  
Variables store temporary data, but files keep data **even after the program ends**.  
Python provides built-in functions to create, read, write, append, and delete files.

---

# <span style="color:#AF7AC5;">📌 Operations Performed on Files</span>
✔ **Create** a new file  
✔ **Write** data into a file  
✔ **Read** data from a file  
✔ **Append** additional data  
✔ **Close** the file after use  

These operations make file handling essential for data storage and processing.

---

# <span style="color:#AF7AC5;">📌 File Opening Modes</span>

| Mode | Meaning | Purpose |
|------|---------|---------|
| `"w"` | Write Mode | Creates a new file or overwrites an existing one |
| `"r"` | Read Mode | Reads the file (file must already exist) |
| `"a"` | Append Mode | Adds new data at the end of the file |
| `"w+"` | Write + Read | Overwrites file and allows reading |
| `"r+"` | Read + Write | Reads first, then writes (does not overwrite) |
| `"a+"` | Append + Read | Appends to file and allows reading |

---

# <span style="color:#AF7AC5;">📌 Combined Example: Write + Read from a Text File</span>

```python
# -----------------------------
# WRITE to a File
# -----------------------------
f = open("sample.txt", "w")        
f.write("Hello! This is a sample file.\n")
f.write("Python file handling is easy.")
f.close()                          


# -----------------------------
# READ from a File
# -----------------------------
f = open("sample.txt", "r")        
content = f.read()                 
f.close()

print(content)

# <span style="color:#AF7AC5;">📘 What is Matplotlib?</span>
Matplotlib is a Python library used to create **basic visualizations** like line graphs, bar graphs, scatter plots, and histograms.  
It provides full control over figure styling.

---

# <span style="color:#AF7AC5;">📘 What is Seaborn?</span>
Seaborn is built on top of Matplotlib.  
It provides **beautiful, statistical, and advanced visualizations** such as box plots, heatmaps, and pairplots with very little code.

---

# <span style="color:#AF7AC5;">📌 Definitions of All Graphs</span>

### 🔹 **Line Graph**
Shows **trends over time** or sequential data connected with lines.

### 🔹 **Bar Graph**
Used to **compare categories** using rectangular bars.

### 🔹 **Scatter Plot**
Shows the **relationship between two variables** using dots.

### 🔹 **Histogram**
Displays the **distribution** of numerical data grouped into bins.

### 🔹 **Box Plot (Seaborn)**
Shows **median, quartiles, and outliers** for statistical comparison.

### 🔹 **Heatmap (Seaborn)**
Displays **matrix-type data** with varying colors showing intensity or value.

### 🔹 **Pairplot (Seaborn)**
Shows **pairwise relationships** between all variables in a dataset.

---

# <span style="color:#AF7AC5;">📌 All Graphs in ONE Merged Code Block</span>

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Data for plots
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
names = ["A", "B", "C"]
values = [40, 60, 80]
data = [10, 20, 22, 25, 30, 40, 45, 50]

# 1. Line Graph
plt.plot(x, y)
plt.title("Line Graph")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()

# 2. Bar Graph
plt.bar(names, values)
plt.title("Bar Graph")
plt.show()

# 3. Scatter Plot
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()

# 4. Histogram
plt.hist(data, bins=5)
plt.title("Histogram")
plt.show()

# ----------- Seaborn Plots -----------

# 5. Box Plot
sns.boxplot(data=data)
plt.title("Box Plot")
plt.show()

# 6. Heatmap
df = pd.DataFrame({
    "A":[1,2,3],
    "B":[4,5,6],
    "C":[7,8,9]
})
sns.heatmap(df, annot=True)
plt.title("Heatmap")
plt.show()

# 7. Pairplot
df2 = sns.load_dataset("iris")
sns.pairplot(df2)
plt.show()
