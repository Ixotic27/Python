```python
class Calculator:
    
    def sum(self,*nums):
        if not nums:
            return 0
        total=0
        for i in nums:
            total+=i
        return total
    
    def diff(self,*nums):
        if not nums:
            return 0
        dif=nums[0]

        for i in nums[1:]:
            dif-=i
        return dif

c=Calculator()

print(c.sum(1,2,3))
print(c.sum(1,2,3,4,5))
print(c.diff(5,4,3))

```

```python
class Animal:
    def sound(self):
        print("Animal makes sound!")

class Dog(Animal): #Single inheritance
    def dogSound(self):
        print("Woof!")

d=Dog()
d.sound()
d.dogSound()
```

```python
class Animal:
    def sound(self):
        print("Animal makes sound!")

class Dog(Animal): #Single inheritance
    def dogSound(self):
        print("Woof!")
    
class Cat(Animal): #Hierarchical
    def catSound(self):
        print("Meow!")

d=Dog()
c=Cat()

d.sound()
d.dogSound()
c.sound()
c.catSound()
```

```python
class A:
    def showA(self):
        print("From class A")

class B(A): #Multi level
    def showB(self):
        print("From class B")

class C(B):
    def showC(self):
        print("From Class C")

c=C()
c.showA()
c.showB()
c.showC()
```

```python
#Function overriding
class A:
    def show(self):
        print("A")

class B:
    def show(self):
        print("B")

for obj in[A(),B()]:
    obj.show()
```

```python
#Operator OVerloading
class Number:
    def __init__(self,value):
        self.value=value
    
    def __add__(self,other):
        return self.value+other.value

s1=Number(10)
s2=Number(20)
print(s1+s2)
```

```python
import numpy as np #IQR METHOD
data = np.array([12, 14, 15, 16, 17, 90])

Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
IQR=Q3-Q1

lower_limit=Q1-(1.5*IQR)
upper_limit=Q3+(1.5*IQR)

outliers=data[(data<lower_limit) | (data>upper_limit)]
print("Outliers: ",outliers)
```

```python
from scipy.stats import zscore
import numpy as np

data = np.array(list(range(1, 101)) + [1000])

z_score=zscore(data)

outliers=data[abs(z_score)>3]
print("Outliers: ",outliers)
```
